```python
score_columns=["math score","reading score","writing score"]
```

```python
for columns in score_columns:
   mean=df[columns].mean()
   median=df[columns].median()
   mode=(df[columns].mode()[0])
```

```python
print("mean_score=",mean)
print("median_score=",median)
print("mode_score=",mode)
```

```python
for columns in score_columns:
    mean=df[columns].mean()
    median=df[columns].median()
    mode=df[columns].mode()
    std=df[columns].std()
    Q1=df[columns].quantile(0.25)
    Q3=df[columns].quantile(0.75)
    print("mean:",mean)
    print("median:",median)
    print("mode:",mode)
    print("Q1:",Q1)
    print("Q3:",Q3)
```

```python
for columns in ["math score","reading score","writing score"]:
    IQR = Q3-Q1
    LOWER=Q1-1.5*IQR
    UPPER=Q3+1.5*IQR
    OUTLIERS=df[
        (df[columns]<LOWER) |
        (df[columns]>UPPER)]
    print("outliers in",columns)
    print(OUTLIERS[columns]) 
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Percentage": [95, 82, 65, 52, 40]
}

df = pd.DataFrame(data)

def get_performance(percentage):
    if percentage >= 90:
        return "Excellent"
    elif percentage >= 75:
        return "Very Good"
    elif percentage >= 60:
        return "Good"
    elif percentage >= 50:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Percentage"].apply(get_performance)

print(df[["Percentage", "Performance"]])
```

```python
import pandas as pd

df = pd.DataFrame({
    "Math": [85, 90],
    "Science": [80, 95],
    "English": [75, 85]
})

score_columns = ["Math", "Science", "English"]

df["total marks"] = df[score_columns].sum(axis=1, numeric_only=True)

print(df)
```

```python
df["Percentage"] = (df["total marks"] / 300) * 100
```
